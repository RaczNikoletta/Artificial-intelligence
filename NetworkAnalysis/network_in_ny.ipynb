```python
import networkx as nx
import matplotlib.pyplot as plt

# 1. Beolvasás
G = nx.read_graphml("path_to_file/new_york_streets.graphml")

# 2. Alapstatisztikák kiíratása
print("Csomópontok száma (Node-ok):", G.number_of_nodes())
print("Élek száma:", G.number_of_edges())
```

```python
degrees = [d for n, d in G.degree()]
plt.hist(degrees, bins=30, color='blue', alpha=0.7)
plt.title("New York úthálózat fokszámeloszlása")
plt.xlabel("Fokszám (Degree)")
plt.ylabel("Gyakoriság")
plt.savefig("fokszam.png")
plt.show()
```
